## 1. Overview
The Vancouver City FC has been facing a decline in revenue across three pillars: Stadium operations, Fanbase engagement, and Merchandise sales, as well as issues engaging international audiences. We aim to optimize these sources of revenue to modernize Vancouver City FC's business model.

- **Project**: Vancouver City FC Revenue Insight
- **Source**: Bolt Case Package
- **Date Range**: Dec. 2023 - Dec. 2024

---

## 2. Data Sources

| Source Name       | Description                          | Format |
|-------------------|--------------------------------------|--------|
| `fanbase_clean.csv`     | Member profiles and demographics     | CSV    |
| `cleaned_merch.csv`       | Purchase history and item categories | CSV    |
| `clean_stadium.csv`     | Stadium operations incl. revenue sources | CSV    |

## 3. Data Dictionary

### Fanbase Engagement (`fanbase_clean.csv`) 70000 records

| Column Name         | Type    | Description                                |
|---------------------|---------|--------------------------------------------|
| `Member_ID`         | integer  | Unique membership identifier (ID)                |
| `Customer_Age_Group`| string (Ordinal)   | Age bracket    |
| `Customer_Region`   | string  (Categorical) | Country or broader region the member is from   |
| `Seasonal_Pass`     | boolean | Whether customer holds a seasonal pass   |
| `Games_Attended` | integer | Number of games attended|


![Games attended by season pass](img/density_games.png)

Binary behaviour of number of games attended when grouped by season pass holder status

### Stadium Operations( `clean_stadium.csv`) 144 records

| Column Name     | Type    | Description                                |
|-----------------|---------|--------------------------------------------|
| `Month` | integer (Ordinal)  | The month within the 2024 Calendar Year    |
| `Revenue`       | integer   | Revenue amount generated by assigned source during that month |
| `Source`     | string (Categorical) | Specific source of revenue or cost   |



### Merchandise Purchase ( `cleaned_merch.csv`) 67877 records

| Column Name           |  Type   | Description                                      |
|-----------------------|-------------|--------------------------------------------------|
| `Product_ID   `         | Integer  (Categorical)   | Unique product identifier                        |
| `Barcode `              | Integer     | The barcode (e.g., UPC/EAN) associated with the exact SKU sold to an individual customer; this is the number scanned at checkout. |
| `Item_Category   `      | Categorical | Type of merchandise                              |
| `Item_Name `            | Categorical | Specific item name                               |
| `Size`                  | Categorical | Item size                                         |
| `Unit_Price `           | Integer  (Ordinal)   | Price per unit (CAD)                             |
| `Customer_Age_Group`    | Categorical | Estimated age bracket of customer, may differ from purchaser age  |
| `Customer_Region `      | Categorical | Domestic vs international                        |
| `Promotion `            | Boolean     | Indicates whether a specific promotional campaign led to the purchase. If the customer engaged with a tracked promotion (e.g., clicked an ad) and then purchased, this is TRUE; otherwise, it is FALSE. |
| `Channel `              | Categorical | Sales channel                                    |
| `Selling_Date`          | Date        | Date and time of transaction                              |
| `Arrival_Date`          | Date        | For online orders, the date and time the product was delivered to the customer. This field is left blank (or NA) for in-store purchases  |
